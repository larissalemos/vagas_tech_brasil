```python
# Função para verificar a Categoria da vaga

def categorizar_vaga(Titulo):
    if 'DESENVOLVEDOR' in Titulo or 'SOFTWARE' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'FRONT' in Titulo or 'BACK' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'PYTHON' in Titulo or 'GOLANG' in Titulo or 'STACK' in Titulo or 'JAVA' in Titulo or 'MOBILE' in Titulo or 'DESARROLLADOR' in Titulo or 'NODE' in Titulo or 'REACT' in Titulo or 'ETL' in Titulo or 'DEV' in Titulo:
        return 'DESENVOLVIMENTO'
    elif 'DATA' in Titulo or 'DADOS' in Titulo or 'DBA' in Titulo or ' ML ' in Titulo or 'POWER' in Titulo or 'INTELLIGENCE' in Titulo or 'BI ANALYST' in Titulo or 'BI ENGINEER' in Titulo or ' BI ' in Titulo or 'ANALYTICS' in Titulo or ' IA ' in Titulo or 'ARTIFICIAL' in Titulo or 'MACHINE LEARNING' in Titulo or 'APRENDIZADO DE MÁQUINA' in Titulo or 'BUSINESS' in Titulo or 'DADO' in Titulo or 'DATOS' in Titulo or 'INTELIGÊNCIA' in Titulo or 'IA' in Titulo or 'BANCO DE OS' in Titulo:
        return 'DADOS/IA/ML'
    elif 'DESIGN' in Titulo or 'UX' in Titulo or 'UI' in Titulo:
        return 'DESIGN/UX'
    elif 'DEVOPS' in Titulo or 'DEVSECOPS' in Titulo or 'RISCO' in Titulo or 'QUALITY' in Titulo or 'QUALIDADE' in Titulo:
        return 'DEVOPS/QA'
    elif 'SEGURANÇA' in Titulo or 'CIBER' in Titulo or 'CYBER' in Titulo or 'SECURITY' in Titulo or 'SEGURANCA' in Titulo or 'MALWARE' in Titulo:
        return 'SEGURANÇA'
    elif 'MARKETING' in Titulo or 'MKT' in Titulo or 'SOCIAL' in Titulo or 'CONTEUDO' in Titulo or 'MARKETER' in Titulo or 'COMMERCE' in Titulo or 'SOCIAIS' in Titulo or 'MÍDIA' in Titulo or 'MEDIA' in Titulo or 'GROWTH' in Titulo:
        return 'MARKETING'
    elif 'SISTEMAS' in Titulo or 'CLOUD' in Titulo or 'NUVEM' in Titulo or 'INFRA' in Titulo or 'ARCHITECT' in Titulo or 'CONNECTIVITY' in Titulo or 'REDE' in Titulo or 'SYSTEM' in Titulo:
        return 'INFRA/CLOUD'
    elif 'LEAD' in Titulo or 'PRODUCT' in Titulo or 'SCRUM' in Titulo or 'PRODUTO' in Titulo or 'AGILE' in Titulo or 'ÁGIL' in Titulo:
        return 'LIDERANÇA'
    elif 'CHATBOT' in Titulo or 'RPA' in Titulo:
        return 'RPA'
    elif 'CRM' in Titulo or 'CRO' in Titulo or 'SALESFORCE' in Titulo:
        return 'CRM/CRO'
    elif 'SALES' in Titulo or 'VENDAS' in Titulo or 'VENDEDOR' in Titulo:
        return 'VENDAS'
    elif 'CUSTOMER' in Titulo or 'CX' in Titulo or 'CLIENTE' in Titulo:
        return 'CX'
    elif ' TI ' in Titulo or 'TECNOLOGIA' in Titulo or 'INFORMAÇÃO' in Titulo or 'TECHNOLOGY' in Titulo or ' IT ' in Titulo:
        return 'GERAL'
    elif 'TEST' in Titulo or 'QA' in Titulo:
        return 'TESTE'
    elif 'SRE' in Titulo or 'RELIABILITY' in Titulo:
        return 'SRE'
    elif 'ERP' in Titulo:
        return 'ERP' 

    # Se nenhuma correspondência for encontrada, você pode retornar algo como 'outro' ou None
    return 'OUTRO'

# Aplicando a função à coluna 'título' para criar a coluna 'Categoria'
df['Categoria'] = df['Titulo'].apply(categorizar_vaga)

# Exibindo o DataFrame resultante
df.head(3)
```

```python
# Função para verificar a sub-Categoria da vaga

def sub_categorizar_vaga(Titulo):
    if ('DATA' in Titulo or 'DADOS' in Titulo or 'DATOS' in Titulo) and ('ENGINEER' in Titulo or 'ENGENHEIR' in Titulo or 'ENGENHARIA' in Titulo):
        return 'ENGENHARIA DE DADOS'
    elif ('DATA' in Titulo or 'DADOS' in Titulo or 'DATOS' in Titulo) and ('SCIENCE' in Titulo or 'CIÊNCIA' in Titulo or 'CIENTISTA' in Titulo or 'SCIENTIST' in Titulo):
        return 'CIÊNCIA DE DADOS'
    elif ('DATA' in Titulo or 'DADOS' in Titulo or 'DATOS' in Titulo or ' BI ' in Titulo) and ('ANALYST' in Titulo or 'ANÁLISE' in Titulo or 'ANALISE' in Titulo or 'ANALYSIS' in Titulo):
        return 'ANALISTA DE DADOS'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('STACK' in Titulo):
        return 'FULL STACK'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('FRONT' in Titulo):
        return 'FRONT END'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('BACK' in Titulo):
        return 'BACK END'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('MOBILE' in Titulo):
        return 'MOBILE'

    # Se nenhuma correspondência for encontrada, você pode retornar algo como 'outro' ou None
    return 'OUTRO'

# Aplicando a função à coluna 'título' para criar a coluna 'Categoria'
df['Sub_Categoria'] = df['Titulo'].apply(sub_categorizar_vaga)

# Exibindo o DataFrame resultante
df.head(3)
```

```python
# Função para verificar a senioridade da vaga

def nivel_vaga(Titulo):
    if 'ESTAGIO' in Titulo or 'ESTÁGIO' in Titulo or 'ASSISTENTE' in Titulo or 'JUNIOR' in Titulo or 'JÚNIOR' in Titulo or 'JR' in Titulo:
        return 'JUNIOR'
    elif 'PLENO' in Titulo or 'PL' in Titulo or 'SENIOR' in Titulo:
        return 'PLENO-SENIOR'
    elif 'ANALISTA' in Titulo or 'ANALYST' in Titulo:
        return 'ANALISTA'
    elif 'CONSULTOR' in Titulo or 'ENGINEER' in Titulo or 'CONSULTANT' in Titulo or 'ESPECIALISTA' in Titulo or 'SPECIALIST' in Titulo or 'EXECUTIVE' in Titulo:
        return 'LEAD'

    # Se nenhuma correspondência for encontrada, você pode retornar algo como 'outro' ou None
    return 'OUTRO'

# Aplicando a função à coluna 'título' para criar a coluna 'nível'
df['nivel'] = df['Titulo'].apply(nivel_vaga)

# Exibindo o DataFrame resultante
df
```

```python
# Função para verificar se a vaga é remota, híbrida ou presencial

def tipo_loc(descricao):
    if 'REMOTO' in descricao or 'HOME' in descricao:
        return 'REMOTO'
    elif 'HÍBRIDO' in descricao or 'HIBRIDO' in descricao:
        return 'HÍBRIDO'
    elif 'PRESENCIAL' in descricao:
        return 'PRESENCIAL'

    # Se nenhuma correspondência for encontrada, você pode retornar algo como 'outro' ou None
    return 'OUTRO'

# Aplicando a função à coluna 'Descricao' para criar a coluna 'tipo_loc'
df['tipo_loc'] = df['Descricao'].apply(tipo_loc)

# Exibindo o DataFrame resultante
df
```

### Cálculos e análises

```python
# Calculando a média do número de candidaturas por categoria de vaga

Calc_Cand_Vaga = df.groupby('Categoria').agg(
    Quantidade=('Categoria', 'count'),
    Soma_Candidaturas=('Candidaturas', 'sum')
).reset_index()
Calc_Cand_Vaga['Cand_Vaga'] = Calc_Cand_Vaga['Soma_Candidaturas'] / Calc_Cand_Vaga['Quantidade']
Calc_Cand_Vaga = Calc_Cand_Vaga.sort_values(by='Cand_Vaga', ascending=True)

# Exibir o DataFrame resultante
print(Calc_Cand_Vaga)
```

```python
# Calculando a média do número de candidaturas por sub-categoria de vaga

Calc_Cand_Sub = df.groupby('Sub_Categoria').agg(
    Quantidade=('Sub_Categoria', 'count'),
    Soma_Candidaturas=('Candidaturas', 'sum')
).reset_index()
Calc_Cand_Sub['Cand_Vaga'] = Calc_Cand_Sub['Soma_Candidaturas'] / Calc_Cand_Sub['Quantidade']
Calc_Cand_Sub = Calc_Cand_Sub.sort_values(by='Cand_Vaga', ascending=True)

# Exibir o DataFrame resultante
print(Calc_Cand_Sub)
```
