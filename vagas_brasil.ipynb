```python
# Preenchendo os estados em branco com base nas cidades de acordo com o arquivo 'cidades_estados_brasil.csv'
def preencher_estado(row):
    if pd.isnull(row['Estado']):
        cidade = row['Cidade']
        estado = cidades_estados[cidades_estados['Cidade'] == cidade]['Estado'].values
        if estado.size > 0:
            return estado[0]
    return row['Estado']
df['Estado'] = df.apply(preencher_estado, axis=1)
# Verificando se ainda há valores nulos na coluna 'Estado'
print('Valores nulos na coluna Estado após preenchimento:')
print(df['Estado'].isnull().sum())
```

```python
# Corrigindo nomes de cidades
def corrigir_cidade(cidade):
    if cidade == 'Brazil':
        return 'Não especificado'
    elif cidade == 'Brasil':
        return 'Não especificado'
    elif cidade == 'Goiania':
        return 'Goiânia'
    elif cidade == 'Cuiaba':
        return 'Cuiabá'
    elif cidade == 'Florianopolis':
        return 'Florianópolis'
    elif cidade == 'EMEA':
        return 'Não especificado'
    elif cidade == 'Brasilia':
        return 'Brasília'
    elif cidade == 'Belem':
        return 'Belém'
    else:
        return cidade
df['Cidade'] = df['Cidade'].apply(corrigir_cidade)
```

### Categorizando vagas de acordo com títulos

```python
# Identificando valores únicos da coluna Titulo
valores_unicos_Titulo = df['Titulo'].unique()
valores_unicos_Titulo
```

```python
# Função para categorizar a vaga com base no Título
def categorizar_vaga(titulo):
    titulo = str(titulo).upper()
    if any(keyword in titulo for keyword in ["DESENVOLVEDOR", "SOFTWARE", "DEVELOPER", "PROGRAMADOR", "FRONT", "BACK", "DESENVOLVIMENTO", "PYTHON", "GOLANG", "STACK", "JAVA", "MOBILE", "DESARROLLADOR", "NODE", "REACT", "DEV"]):
        return "DESENVOLVIMENTO"
    elif any(keyword in titulo for keyword in ["NEGÓCIOS", "ANALISTA DE PROJETOS", "CUSTOMER SUCCESS", "OPERAÇÕES", "OPERATIONS", "CRM", "CX", "BUSINESS ANALYST", "SISTEMAS", "ANALISTA FINANCEIRO", "GROWTH", "MELHORIA CONTÍNUA", "ANALISTA DE SISTEMAS", "ANALISTA DE NEGÓCIOS", "LOGÍSTICA", "ERP", "SAP", "ORACLE ERP", "TOTVS", "DYNAMICS"]):
        return "ANALISTAS DE NEGÓCIOS"
    elif any(keyword in titulo for keyword in ["TI", "IT", "ANALISTA DE TECNOLOGIA", "ANALISTA DE GESTÃO DA INFORMAÇÃO", "ANALISTA DE TECNOLOGIA", "ANALISTA DE SISTEMAS", "SYSTEM"]):
        return "ANALISTA DE SISTEMAS"
    elif any(keyword in titulo for keyword in ["DATA", "DADOS", "DBA", "BANCO DE DADOS", "SQL", "ML", "MACHINE LEARNING", "POWER BI", "INTELLIGENCE", "INTELIGÊNCIA", "QUALITATIVE", "DADO", "ANALISTA DE NEGÓCIO", "ANALISTA DE GROWTH", "ESTATÍSTICA", "STATISTICS", "ETL", "INTELIGENCE", "DASHBOARD", "BI ANALYST", "ANALISTA DE BI", "ANALYTICS", "BI ENGINEER", "AI ", "BI ", " IA"]):
        return "DADOS/BI"
    elif any(keyword in titulo for keyword in ["QA", "TESTE", "QUALIDADE", "QUALITY", "TEST"]):
        return "QUALIDADE/TESTES"
    elif any(keyword in titulo for keyword in ["SEGURANÇA", "CYBER", "RISCO", "SECURITY", "FRAUDE"]):
        return "SEGURANÇA DA INFORMAÇÃO"
    elif any(keyword in titulo for keyword in ["CLOUD", "ARQUITETO", "ARCHITECT", "DEVOPS", "TERRAFORM", "ANSIBLE", "INFRAESTRUTURA", "RELIABILITY", "SRE", "INFRA", "INFRASTRUCTURE", "KUBERNET", "DOCKER", "AWS", "AZURE", "GCP", "CICD", "CONTÊINERES", "CONTAINERIZATION", "VIRTUALIZAÇÃO", "KUBERNETES", "ORACLE", "REDES", "NETWORK", "SDN", "ROUTER", "SWITCH", "FIREWALL"]):
        return "CLOUD/DEVOPS/INFRA"
    elif any(keyword in titulo for keyword in ["SUPORTE", "HELP DESK"]):
        return "SUPORTE TÉCNICO"
    elif any(keyword in titulo for keyword in ["UX", "UI", "DESIGNER", "DESIGN", "USER EXPERIENCE", "USER INTERFACE", "PROTOTIPAGEM", "FIGMA", "UX DESIGNER", "UI DESIGNER", "UX/UI", "DESIGNER GRÁFICO", "DESIGNER DE INTERAÇÃO", "DESIGNER DE PRODUTO"]):
       return "DESIGN/UX/UI"
    elif any(keyword in titulo for keyword in ["GERENTE DE PROJETOS", "PO", "PRODUCT OWNER", "GERENTE DE PRODUTO", "GERENTE DE TI", "COORDENADOR", "TECH LEAD", "SCRUM", "PROJECT MANAGER", "AGILE", "PMO", "PRODUCT MANAGER"]):
        return "LIDERANÇA"
    elif any(keyword in titulo for keyword in ["MARKETING", "COMUNICAÇÃO", "SOCIAL MEDIA", "SEO", "SEM", "MÍDIA", "CONTEÚDO"]):
        return "MARKETING/DIGITAL"
    elif any(keyword in titulo for keyword in ["BANCO DE TALENTOS"]):
        return "BANCO DE TALENTOS"
    else:
        return "OUTROS"

df["Categoria"] = df["Titulo"].apply(categorizar_vaga)
print("Vagas categorizadas com base no título.")
print(df[["Titulo", "Categoria"]].head())
```

### Criando sub-categorias

```python
# Função para verificar a sub-Categoria da vaga

def sub_categorizar_vaga(Titulo):
    if ('DATA' in Titulo or 'DADOS' in Titulo or 'DATOS' in Titulo) and ('ENGINEER' in Titulo or 'ENGENHEIR' in Titulo or 'ENGENHARIA' in Titulo):
        return 'ENGENHARIA DE DADOS'
    elif ('DATA' in Titulo or 'DADOS' in Titulo or 'DATOS' in Titulo) and ('SCIENCE' in Titulo or 'CIÊNCIA' in Titulo or 'CIENTISTA' in Titulo or 'SCIENTIST' in Titulo):
        return 'CIÊNCIA DE DADOS'
    elif ('DATA' in Titulo or 'DADOS' in Titulo or 'DATOS' in Titulo or ' BI ' in Titulo) and ('ANALYST' in Titulo or 'ANÁLISE' in Titulo or 'ANALISE' in Titulo or 'ANALYSIS' in Titulo):
        return 'ANALISTA DE DADOS'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('STACK' in Titulo):
        return 'FULL STACK'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('FRONT' in Titulo):
        return 'FRONT END'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('BACK' in Titulo):
        return 'BACK END'
    elif ('DESENVOLVEDOR' in Titulo or 'DEVELOPER' in Titulo or 'PROGRAMADOR' in Titulo or 'DESENVOLVIMENTO' in Titulo or 'DEV' in Titulo) and ('MOBILE' in Titulo):
        return 'MOBILE'

    # Se nenhuma correspondência for encontrada, você pode retornar algo como 'outro' ou None
    return 'OUTRO'

# Aplicando a função à coluna 'título' para criar a coluna 'Categoria'
df['Sub_Categoria'] = df['Titulo'].apply(sub_categorizar_vaga)

# Exibindo o DataFrame resultante
df.head(3)
```

### Verificando modelo de trabalho de acordo com descrição

```python
# Função para verificar se a vaga é remota, híbrida ou presencial

def modelo(descricao):
    if 'REMOTO' in descricao or 'HOME' in descricao or 'REMOTE' in descricao or 'TELETRABALHO' in descricao:
        return 'REMOTO'
    elif 'HÍBRIDO' in descricao or 'HIBRIDO' in descricao or 'HÍBRIDA' in descricao or 'HYBRID' in descricao:
        return 'HÍBRIDO'
    elif 'PRESENCIAL' in descricao or 'ON-SITE' in descricao or 'IN-PERSON' in descricao:
        return 'PRESENCIAL'

    # Se nenhuma correspondência for encontrada, retornar OUTRO
    return 'OUTRO'

# Aplicando a função à coluna 'Descricao' para criar a coluna 'tipo_loc'
df['modelo'] = df['Descricao'].apply(modelo)

# Exibindo o DataFrame resultante
df
```

### Análise de Candidaturas

```python
# Calculando a média do número de candidaturas por categoria de vaga

Calc_Cand_Vaga_cat = df.groupby('Categoria').agg(
    Quantidade=('Categoria', 'count'),
    Soma_Candidaturas=('Candidaturas', 'sum')
).reset_index()
Calc_Cand_Vaga_cat['Cand_Vaga'] = Calc_Cand_Vaga_cat['Soma_Candidaturas'] / Calc_Cand_Vaga_cat['Quantidade']
Calc_Cand_Vaga_cat = Calc_Cand_Vaga_cat.sort_values(by='Cand_Vaga', ascending=True)

# Exibir o DataFrame resultante
print(Calc_Cand_Vaga_cat)
```

```python
# Calculando o número de candidaturas por vaga por categoria e nível de senioridade
Calc_Cand_Vaga_cat = df.groupby(['Categoria', 'Senioridade']).agg(
    Quantidade_Vagas=('Categoria', 'count'),  # Soma o número de vagas
    Soma_Candidaturas=('Candidaturas', 'sum')
).reset_index()

# Calculando candidaturas por vaga (média)
Calc_Cand_Vaga_cat['Cand_Vaga'] = Calc_Cand_Vaga_cat['Soma_Candidaturas'] / Calc_Cand_Vaga_cat['Quantidade_Vagas']

# Ordenando por candidaturas por vaga (do maior para o menor)
Calc_Cand_Vaga_cat = Calc_Cand_Vaga_cat.sort_values(by='Cand_Vaga', ascending=False)

# Formatando para melhor visualização
Calc_Cand_Vaga_cat['Cand_Vaga'] = Calc_Cand_Vaga_cat['Cand_Vaga'].round(2)

# Exibir o DataFrame resultante
Calc_Cand_Vaga_cat
```

### Sankey de categorias e senioridade

```python
# Sankey com porcentagens relativas de senioridade por categoria
df_sankey_senioridade = df.groupby(["Categoria", "Senioridade"]).size().reset_index(name="Count")

# Calcular totais por categoria
totais_categoria = df_sankey_senioridade.groupby("Categoria")["Count"].sum().reset_index()
df_sankey_senioridade = df_sankey_senioridade.merge(totais_categoria, on="Categoria", suffixes=('', '_total'))

# Calcular porcentagens relativas
df_sankey_senioridade["Porcentagem"] = (df_sankey_senioridade["Count"] / df_sankey_senioridade["Count_total"] * 100).round(2)

unique_categorias_sankey = df_sankey_senioridade["Categoria"].unique()
unique_senioridades_sankey = df_sankey_senioridade["Senioridade"].unique()

all_nodes_sankey = np.concatenate((unique_categorias_sankey, unique_senioridades_sankey))
label_map_sankey = {label: i for i, label in enumerate(all_nodes_sankey)}

# Modificar labels para incluir informações
node_labels = []
for node in all_nodes_sankey:
    if node in unique_categorias_sankey:
        total_cat = df_sankey_senioridade[df_sankey_senioridade["Categoria"] == node]["Count_total"].iloc[0]
        node_labels.append(f"{node}<br>({total_cat} vagas)")
    else:
        node_labels.append(node)

source_sankey = [label_map_sankey[cat] for cat in df_sankey_senioridade["Categoria"]]
target_sankey = [label_map_sankey[sen] for sen in df_sankey_senioridade["Senioridade"]]
value_sankey = df_sankey_senioridade["Porcentagem"].tolist()

fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=node_labels,
    ),
    link=dict(
        source=source_sankey,
        target=target_sankey,
        value=value_sankey,
        hovertemplate='%{source.label} → %{target.label}<br>%{value}% da categoria<extra></extra>',
    )
)]).update_layout(
    title_text="Distribuição de Senioridade por Categoria (em %)",
    font_size=10
)
fig.show()
```
